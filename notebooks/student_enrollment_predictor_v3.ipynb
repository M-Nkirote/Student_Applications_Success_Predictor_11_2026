## Target definition

Each stage of 'Most Advanced Stage is placed into one of three outcome groups:

| Group | Stages | Used for |
|---|---|---|
| **Won** (target = 1) | Student Enrolled · Enrolled, but Dropped Out · Enrolled, but no Commission | Training & evaluation |
| **Lost** (target = 0) | Application Withdrawn · Application Rejected · Offer Declined · Visa Application Withdrawn · Visa Application Rejected · Did Not Enroll (after paying deposit) · Did Not Enroll (after visa issued) · Closed-Lost to Competition · Not Interested | Training & evaluation |
| **Open** (no label yet) | Open Opportunity · Awaiting Documents · Course Decided · Application Ready for Submission · Application Done · Conditional Offer Received/Accepted · Offer Received/Accepted · Visa Application Submitted · Visa Issued · Deposit Paid · Future Intake | **Scored** by the final model |

```python
WON_STAGES = ['Student Enrolled', 'Enrolled, but Dropped Out', 'Enrolled, but no Commission']

LOST_STAGES = ['Application Withdrawn', 'Application Rejected', 'Offer Declined',
               'Visa Application Withdrawn', 'Visa Application Rejected',
               'Did Not Enroll (after paying deposit)', 'Did Not Enroll (after visa issued)',
               'Closed-Lost to Competition', 'Not Interested']

EARLY_OPEN_STAGES = ['Open Opportunity', 'Awaiting Documents', 'Course Decided',
                     'Application Ready for Submission', 'Application Done']
LATE_OPEN_STAGES = ['Conditional Offer Received', 'Conditional Offer Accepted', 'Offer Received', 'Offer Accepted',
                    'Visa Application Submitted', 'Visa Issued', 'Deposit Paid', 'Future Intake']

STALE_MONTHS = 12  

# Every stage in the data must belong to exactly one group
all_known = WON_STAGES + LOST_STAGES + EARLY_OPEN_STAGES + LATE_OPEN_STAGES
unknown_stages = set(df['Most Advanced Stage'].unique()) - set(all_known)
assert not unknown_stages, f'Stages not mapped to a group: {unknown_stages}'
```

```python
# Step 1:basic outcome group
df['outcome'] = 'open'
df.loc[df['Most Advanced Stage'].isin(WON_STAGES), 'outcome'] = 'won'
df.loc[df['Most Advanced Stage'].isin(LOST_STAGES), 'outcome'] = 'lost'

# Step 2: stale early-stage open applications to 'lost'
df['age_months_at_snapshot'] = (SNAPSHOT_DATE - df['Created At']).dt.days / 30.44
if STALE_MONTHS is not None:
    is_stale = (df['Most Advanced Stage'].isin(EARLY_OPEN_STAGES)) & (df['age_months_at_snapshot'] > STALE_MONTHS)
    df.loc[is_stale, 'outcome'] = 'lost (stale)'

# Step 3: binary target (missing for open applications)
df['target'] = np.where(df['outcome'] == 'won', 1, np.where(df['outcome'].str.startswith('lost'), 0, np.nan))

print(df['outcome'].value_counts())
```

### Fix duplicate enrolments

A student can only enrol once, yet some students have several *won* applications. Two possible causes:
- **Duplicate records** - the same enrolment entered twice (same institution and same intake).
- **Different institution or intake** - e.g. a transfer or a re-enrolment after dropping out.

In both cases the business rule says only one enrolment is real, and counting all of them double-counts successes. If most cases turn out to be *Student Enrolled* at **different** institutions, that suggests the CRM marks several of a student's applications as enrolled.

Rule applied:
- **Identifiable** - exactly one of the student's enrolled applications has an 'Enrolled Date' → keep that one as the enrolment and remove the other enrolled copies.
- **Ambiguous** - none or several have an 'Enrolled Date' , we cannot tell which enrolment is real, so **all applications of that student are removed**.

Why not simply keep the earliest (or latest) record? Because then the *position* of the enrolment among a student's applications is decided by our rule, not by reality - and any feature that depends on that position (such as the number of earlier applications) would learn the rule. For example, always keeping the earliest record would make every enrolment a student's first application, and 'prior_applications' would become an artificial "repeat applicants never enrol" signal. Removed records are not relabelled as "lost", because their true outcome is unknown.

```python
won_rows = df[df['outcome'] == 'won']
multi_won_ids = won_rows.loc[won_rows['Student ID'].duplicated(keep=False), 'Student ID'].unique()
multi_won = won_rows[won_rows['Student ID'].isin(multi_won_ids)]
print(f'Students with more than one enrolled application: {len(multi_won_ids)} ({len(multi_won)} applications)')

# Per student: do all enrolled applications share the same institution and intake?
dup_check = multi_won.groupby('Student ID').agg(
    enrolled_applications=('Opportunity ID', 'size'),
    n_institutions=('Institution Name', 'nunique'),
    n_intakes=('Correct Intake', 'nunique'),
    n_with_enrolled_date=('Enrolled Date', 'count'),
)
dup_check['type'] = np.where((dup_check['n_institutions'] <= 1) & (dup_check['n_intakes'] <= 1),
                             'duplicate record', 'different institution/intake')
print(dup_check['type'].value_counts())
dup_check['enrolled_applications'].value_counts().sort_index().rename('students with this many enrolled applications')
```

```python
# Which enrolled stages appear in each type? (e.g. 'Enrolled, but Dropped Out' + 'Student Enrolled' suggests a transfer)
multi_won_typed = multi_won.merge(dup_check[['type']], left_on='Student ID', right_index=True)
pd.crosstab(multi_won_typed['Most Advanced Stage'], multi_won_typed['type'])
```

```python
# How many enrolled applications of each student have an Enrolled Date?
enrolment_dated_applications = multi_won.groupby('Student ID')['Enrolled Date'].count()
identifiable_ids = enrolment_dated_applications[enrolment_dated_applications == 1].index
ambiguous_ids = enrolment_dated_applications[enrolment_dated_applications != 1].index
print(f'Identifiable students (exactly one dated enrolment): {len(identifiable_ids)}')
print(f'Ambiguous students (no or several dated enrolments): {len(ambiguous_ids)}')

# Identifiable: remove the undated enrolled copies
undated_copies = multi_won[multi_won['Student ID'].isin(identifiable_ids) & multi_won['Enrolled Date'].isna()]['Opportunity ID']

# Ambiguous: remove every application of the student
ambiguous_apps = df.loc[df['Student ID'].isin(ambiguous_ids), 'Opportunity ID']

remove_ids = set(undated_copies) | set(ambiguous_apps)
rows_before = len(df)
df = df[~df['Opportunity ID'].isin(remove_ids)].reset_index(drop=True)
print(f'Removed {len(remove_ids)} applications: from {rows_before} to {len(df)} rows')

# Business rule now holds
assert df.loc[df['outcome'] == 'won', 'Student ID'].duplicated().sum() == 0
print('Check passed: every student has at most one enrolled application.')
```

### Training window: only applications whose intake has passed

**Right-censoring problem:** for recent applications, *losses* close quickly (the student says "not interested") while *wins* are still open (the student has not reached the intake yet). So recent closed applications are almost all losses, which drags the recent enrolment rate towards 0% and teaches the model that "recent = bad".

**Fix:** a closed application is used for training/evaluation only if its outcome had a fair chance to happen:
- its intake date is at least `INTAKE_BUFFER_MONTHS` before the snapshot, **or**
- it has no intake date but was created more than `MATURE_MONTHS_NO_INTAKE` months before the snapshot.

Closed applications outside this window are neither trained on nor scored (they already have an outcome). Open applications are still all scored.

```python
INTAKE_BUFFER_MONTHS = 1
MATURE_MONTHS_NO_INTAKE = 12

intake_passed = (df['intake_date'] + pd.DateOffset(months=INTAKE_BUFFER_MONTHS)) <= SNAPSHOT_DATE
old_without_intake = df['intake_date'].isna() & (df['age_months_at_snapshot'] > MATURE_MONTHS_NO_INTAKE)
df['label_mature'] = intake_passed | old_without_intake

closed = df['target'].notna()
print('Closed applications inside the training window :', int((closed & df['label_mature']).sum()))
print('Closed applications outside the window (excluded):', int((closed & ~df['label_mature']).sum()))

# Evidence of censoring: enrolment rate inside vs outside the window
df[closed].groupby('label_mature')['target'].agg(applications='size', enrolment_rate='mean').round(4)
```

## Feature engineering

Every feature below must be **known at (or very near) the moment the application is created**. That is the rule that protects against leakage.

| New feature | How it is built | Idea behind it |
|---|---|---|
| `age_at_application` | `Created At` − `Date Of Birth`, in years; values outside 15–65 → missing | Mature applicants may be more committed / better funded |
| `intake_lead_months` | Months between creation and the target intake | Too little time → can't finish documents/visa; too much → interest fades |
| `intake_month` | Month of the target intake (Jan, May, Sep …) | Main intakes may convert differently |
| `created_month` | Month the application was created | Seasonality of recruitment |
| `prior_applications` | Number of earlier applications by the same student (counted in date order) | **EDA only — not a model feature** (see note below the feature lists) |
| `has_dob`, `has_high_school` | 1 if the field is filled in | Profile completeness signals engagement |
| `studying_in_home_country` | Destination equals the student's country | Local study has no visa barrier |
| `destination_type` | Kenya → *Kenya*, any other destination → *International* |

```python
# Age at application (in years); impossible values become missing
df['age_at_application'] = (df['Created At'] - df['Date Of Birth']).dt.days / 365.25
df.loc[(df['age_at_application'] < 15) | (df['age_at_application'] > 65), 'age_at_application'] = np.nan

# Months between creation and intake; extreme values (data-entry errors) become missing
df['intake_lead_months'] = (df['intake_date'] - df['Created At']).dt.days / 30.44
df.loc[(df['intake_lead_months'] < -12) | (df['intake_lead_months'] > 36), 'intake_lead_months'] = np.nan

# Intake month and creation month as categories (text), because months are cyclical, not ordered numbers
df['intake_month'] = df['intake_date'].dt.strftime('%b').fillna('Undecided')
df['created_month'] = df['Created At'].dt.strftime('%b')

# Creation year / quarter: used for EDA and the temporal check only (not as a model feature, see note below)
df['created_year'] = df['Created At'].dt.year
df['created_quarter'] = df['Created At'].dt.to_period('Q').astype(str)

# Prior applications by the same student, counted in chronological order
df = df.sort_values(['Student ID', 'Created At']).reset_index(drop=True)
df['prior_applications'] = df.groupby('Student ID').cumcount()

# Profile-completeness flags
df['has_dob'] = df['Date Of Birth'].notna().astype(int)
df['has_high_school'] = df['Current Previous High School'].notna().astype(int)

# Studying in the home country
df['studying_in_home_country'] = (df['destination'] == df['Student Country Location']).astype(int)

df[['age_at_application', 'intake_lead_months', 'prior_applications', 'has_dob',
    'has_high_school', 'studying_in_home_country']].describe().T.round(2)
```

```python
# Final feature lists
NUMERIC_FEATURES = ['age_at_application', 'intake_lead_months', 'has_dob', 'has_high_school', 'studying_in_home_country']

CATEGORICAL_FEATURES = ['Lead Source Main Category', 'Deal Owner', 'Student Country Location', 'level_of_study', 'destination', 'destination_type', 'Study Type', 'intake_month', 'created_month', 'Institution Name']

# Features that are filled in / updated as the application progresses
SNAPSHOT_RISK_FEATURES = ['Institution Name', 'has_dob', 'has_high_school', 'age_at_application']

ALL_FEATURES = NUMERIC_FEATURES + CATEGORICAL_FEATURES

# Modelling set = closed applications inside the training window
model_df = df[df['target'].notna() & df['label_mature']].copy()
model_df['target'] = model_df['target'].astype(int)

# Scoring set = open applications (no outcome yet)
open_df = df[df['target'].isna()].copy()

print(f'Modelling set: {len(model_df):,} applications from {model_df["Student ID"].nunique():,} students')
print(f'Open set (to be scored): {len(open_df):,} applications')
print(f'Closed but outside the training window (not used): {int((df["target"].notna() & ~df["label_mature"]).sum()):,}')
```

### Missingness analysis
We look at *how* values are missing: randomly, or in patterns that relate to each other and to the target. Missing data that depends on the outcome is **Missing Not At Random (MNAR)**, and then "is missing" is itself information.

```python
# Missingness matrix: white = missing. Rows sorted by target so patterns by outcome are visible.
missing_cols = ['Date Of Birth', 'Current Previous High School', 'Institution Name','intake_date', 'Student Country Location',
                'Application Submission Date', 'Offer Letter Received Date', 'Deposit Paid Date', 'Visa Issued Date', 'Enrolled Date', 'Application Fee Paid']
msno.matrix(model_df.sort_values('target')[missing_cols], figsize=(14, 6), fontsize=9, sparkline=False)
plt.title('Missingness matrix (rows sorted by target: non-enrolled first, enrolled last)'); plt.show()
```

```python
# Nullity correlation heatmap: do columns tend to be missing together?
msno.heatmap(model_df[missing_cols], figsize=(10, 7), fontsize=9)
plt.title('Nullity correlation'); plt.show()
```

```python
# Is missingness related to the target?  Chi-square test of independence for each column.
# H0: whether the value is missing is independent of enrolment.
rows = []
for col in ['Date Of Birth', 'Current Previous High School', 'Institution Name', 'intake_date']:
    is_missing = model_df[col].isna()
    table = pd.crosstab(is_missing, model_df['target'])
    if table.shape == (2, 2):
        chi2, p, dof, expected = stats.chi2_contingency(table)
    else:
        chi2, p = np.nan, np.nan
    rows.append({'column': col,
                 'enrol_rate_when_missing_%': round(model_df.loc[is_missing, 'target'].mean() * 100, 1),
                 'enrol_rate_when_present_%': round(model_df.loc[~is_missing, 'target'].mean() * 100, 1),
                 'chi2': round(chi2, 1), 'p_value': p})
pd.DataFrame(rows)
```

**Interpretation guide:**
- The post-outcome date columns (submission, offer, deposit, visa, enrolled) are missing **together** and are almost never missing for enrolled students — missingness is **structural**: a date is empty because the stage was never reached. That confirms they are outcome information (leakage), not random gaps.
- If enrolment rates differ strongly between "missing" and "present" for a feature (small p-value), the data is likely **MNAR**. That is why the numeric imputer adds **missing-indicator** columns (`add_indicator=True`) and categorical missing values become their own `"Missing"` category instead of being filled with the mode.

### Bivariate analysis + hypothesis testing

**Categorical features vs target** - enrolment rate per category, with a **chi-square test of independence** and **Cramér's V** (effect size, 0 = no association, 1 = perfect).

- H0: the feature and enrolment are independent.  H1: they are associated.  Significance level α = 0.05.

```python
overall_rate = model_df['target'].mean()

def cramers_v(x, y):
    # Cramér's V effect size from a chi-square test on the contingency table of x and y
    table = pd.crosstab(x, y)
    chi2 = stats.chi2_contingency(table)[0]
    n = table.values.sum()
    r, k = table.shape
    return np.sqrt(chi2 / (n * (min(r, k) - 1))) if min(r, k) > 1 else np.nan

assoc_rows = []
for col in CATEGORICAL_FEATURES:
    x = model_df[col].fillna('Missing')
    table = pd.crosstab(x, model_df['target'])
    chi2, p, dof, _ = stats.chi2_contingency(table)
    assoc_rows.append({'feature': col, 'n_categories': x.nunique(), 'chi2': round(chi2, 1), 'dof': dof,
                       'p_value': p, 'cramers_v': round(cramers_v(x, model_df['target']), 3),
                       'reject_H0 (α=0.05)': p < 0.05})
assoc_table = pd.DataFrame(assoc_rows).sort_values('cramers_v', ascending=False)
assoc_table
```

```python
# Enrolment rate by category (top 10 categories by count) with the overall rate as reference
fig, axes = plt.subplots(2, 4, figsize=(20, 10))
for ax, col in zip(axes.ravel(), plot_cats):
    x = model_df[col].fillna('Missing')
    top = x.value_counts().head(10).index
    rate = model_df[x.isin(top)].groupby(x[x.isin(top)])['target'].mean().sort_values() * 100
    rate.plot.barh(ax=ax, color='seagreen')
    ax.axvline(overall_rate * 100, color='red', linestyle='--', label='overall')
    ax.set_title(f'Enrolment rate (%) by {col}'); ax.set_ylabel(''); ax.legend()
plt.tight_layout(); plt.show()
```

**Numeric features vs target** - **Mann-Whitney U test** (non-parametric; chosen because the normality tests above rejected normality) and the **rank-biserial correlation** as effect size.

- H0: the distribution of the feature is the same for enrolled and non-enrolled applications.  α = 0.05.

```python
num_rows = []
for col in num_cols_eda + ['has_dob', 'has_high_school', 'studying_in_home_country']:
    enrolled = model_df.loc[model_df['target'] == 1, col].dropna()
    not_enrolled = model_df.loc[model_df['target'] == 0, col].dropna()
    u_stat, p = stats.mannwhitneyu(enrolled, not_enrolled, alternative='two-sided')
    rank_biserial = 1 - 2 * u_stat / (len(enrolled) * len(not_enrolled))
    num_rows.append({'feature': col,
                     'median_enrolled': enrolled.median(), 'median_not_enrolled': not_enrolled.median(),
                     'mean_enrolled': round(enrolled.mean(), 3), 'mean_not_enrolled': round(not_enrolled.mean(), 3),
                     'U': u_stat, 'p_value': p, 'rank_biserial': round(rank_biserial, 3),
                     'reject_H0 (α=0.05)': p < 0.05})
pd.DataFrame(num_rows)
```

#### PCA (exploratory)
PCA on the encoded features shows how much structure there is and whether enrolled applications separate from the rest in a low-dimensional view.

```python
# Prepare a numeric matrix: impute + one-hot encode + scale (for visualisation only)
pca_prep = ColumnTransformer([
    ('num', Pipeline([('impute', SimpleImputer(strategy='median')), ('scale', StandardScaler())]), NUMERIC_FEATURES),
    ('cat', Pipeline([('impute', SimpleImputer(strategy='constant', fill_value='Missing')),
                      ('onehot', OneHotEncoder(handle_unknown='ignore', min_frequency=30, sparse_output=False))]),
     [c for c in CATEGORICAL_FEATURES if c not in ['Institution Name', 'Deal Owner']]),
])
X_pca_input = pca_prep.fit_transform(model_df)
X_pca_input = StandardScaler().fit_transform(X_pca_input)   # put one-hot columns on the same scale

pca = PCA(random_state=RANDOM_STATE).fit(X_pca_input)
cum_var = np.cumsum(pca.explained_variance_ratio_)
n_for_80 = int(np.argmax(cum_var >= 0.80) + 1)
print(f'Encoded dimensions: {X_pca_input.shape[1]} | components needed for 80% of variance: {n_for_80}')

fig, axes = plt.subplots(1, 2, figsize=(15, 5))
axes[0].plot(range(1, len(cum_var) + 1), cum_var, marker='.')
axes[0].axhline(0.8, color='red', linestyle='--'); axes[0].set_xlabel('Number of components')
axes[0].set_ylabel('Cumulative explained variance'); axes[0].set_title('PCA scree (cumulative)')

pcs = pca.transform(X_pca_input)[:, :2]
axes[1].scatter(pcs[model_df['target'] == 0, 0], pcs[model_df['target'] == 0, 1], s=5, alpha=0.3, c='grey', label='not enrolled')
axes[1].scatter(pcs[model_df['target'] == 1, 0], pcs[model_df['target'] == 1, 1], s=8, alpha=0.7, c='seagreen', label='enrolled')
axes[1].set_xlabel('PC1'); axes[1].set_ylabel('PC2'); axes[1].set_title('First two principal components'); axes[1].legend()
plt.tight_layout(); plt.show()
```

**Why PCA is *not* used inside the model:** 

(1) variance is spread across many components (no small set of components captures most of it), so compressing would throw away signal; 

(2) the encoded feature space is modest in size, so dimensionality is not a computational problem; 

(3) principal components cannot be explained to an agent ("PC3 is high") whereas original features can; 

(4) tree-based models handle correlated/sparse inputs without needing PCA. PCA is kept as an **exploratory** tool - this is a "thoughtful exclusion" for the technique-justification section.

## Data leakage audit

### Tier 1 - hard leakage (excluded from features)
| Column(s) | Why it leaks |
|---|---|
| `Stage`, `Most Advanced Stage` | They *are* the outcome |
| `Application Submission Date`, `Offer Letter Received Date`, `Application Rejection Date` | Only exist once the application progresses (or fails) |
| `Deposit Paid Date`, `Deposit Amount Paid`, `Date Professional Fee Paid`, `Application Fee Paid` | Payments happen mid/late funnel |
| `Visa Submitted Through Higher Studies`, `Visa Issued Date`, `Enrolled Date` | Late-funnel / outcome events |
| `Enrollment Deferred To Another Intake` | Only known close to the intake |
| Other applications of the same student | A student enrols only once: knowing a sibling application enrolled reveals this one did not → handled by grouping the split by `Student ID` and counting only *prior* applications |

### Tier 2 - "snapshot" risk (kept, but tested)
The data is a **single snapshot** of the CRM: each field holds its *latest* value, not the value at creation. Based on the agency's process, `Study Destination Type` and `Visa Is Required` (filled only once a course is decided and the application starts) were already **dropped** for this reason. `Deal Owner` is **not** a snapshot risk: the agency confirmed applications are never reassigned, so the owner is known from day one. Fields that agents fill in while working the application (date of birth, high school, a specific institution instead of "Undecided") will be more complete for applications that progressed. These can be genuinely useful at scoring time, but part of their apparent power is leakage.

```python
leak_check = pd.DataFrame({
    'has_submission_date': model_df['Application Submission Date'].notna(),
    'has_offer_date': model_df['Offer Letter Received Date'].notna(),
    'has_deposit_date': model_df['Deposit Paid Date'].notna(),
    'application_fee_paid': model_df['Application Fee Paid'].eq('Yes'),
    'visa_through_agency': model_df['Visa Submitted Through Higher Studies'].eq('Yes'),
})
leak_rows = []
for col in leak_check.columns:
    flag = leak_check[col]
    leak_rows.append({'leaky_flag': col,
                      'enrol_rate_if_true_%': round(model_df.loc[flag, 'target'].mean() * 100, 1),
                      'enrol_rate_if_false_%': round(model_df.loc[~flag, 'target'].mean() * 100, 1),
                      'share_of_enrolled_with_flag_%': round(flag[model_df['target'] == 1].mean() * 100, 1)})
pd.DataFrame(leak_rows)
```

```python
# Guard rail: make sure no leakage column slipped into the feature lists
forbidden = set(STAGE_COLS + POST_OUTCOME_COLS + ['target', 'outcome', 'Enrolled Date'])
assert not (set(ALL_FEATURES) & forbidden), f'Leaky columns in features: {set(ALL_FEATURES) & forbidden}'
print('Leakage guard passed: no Tier 1 columns in the feature list.')
```

## Train / test split - grouped by student

`StratifiedGroupKFold` keeps **all applications of a student in the same fold** (group constraint) while keeping the enrolment rate similar across folds (stratification). We take one of 5 folds (≈20%) as the **held-out test set** and do not touch it until Section 14.

```python
X = model_df[ALL_FEATURES]
y = model_df['target']
groups = model_df['Student ID']

splitter = StratifiedGroupKFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)
train_idx, test_idx = next(splitter.split(X, y, groups))

X_train, X_test = X.iloc[train_idx], X.iloc[test_idx]
y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]
groups_train, groups_test = groups.iloc[train_idx], groups.iloc[test_idx]

print(f'Train: {len(X_train):,} rows | {groups_train.nunique():,} students | enrolment rate {y_train.mean():.3%}')
print(f'Test : {len(X_test):,} rows | {groups_test.nunique():,} students | enrolment rate {y_test.mean():.3%}')
```

```python
# Verify: no student appears in both sets
overlap = set(groups_train) & set(groups_test)
print('Students in both train and test:', len(overlap))
assert len(overlap) == 0

# Save the split so it can be reproduced exactly by the scripted pipeline
split_record = pd.DataFrame({'Opportunity ID': model_df['Opportunity ID'],
                             'split': np.where(np.isin(np.arange(len(model_df)), test_idx), 'test', 'train')})
split_record.to_csv(OUTPUT_DIR / 'train_test_split.csv', index=False)
```

## Preprocessing pipeline

All preprocessing lives **inside** a scikit-learn `Pipeline`. This is what makes the work reproducible and leakage-safe: during cross-validation, the imputer medians, the infrequent-category groups and the scaler means are learned **only from the training folds**, and the exact same steps are re-applied at prediction time in the API.

| Step | Numeric features | Categorical features |
|---|---|---|
| Imputation | Median (robust to the skew seen in EDA) **+ missing-indicator columns** (missingness is informative) | Constant `"Missing"` category |
| Encoding | – | One-hot; categories with fewer than `MIN_FREQUENCY` training rows are merged into an *infrequent* bucket; unseen categories at prediction time go to that bucket instead of crashing the API |
| Scaling | `StandardScaler` (needed for logistic regression; harmless for trees) | – |

**Feature set choice:** by default the model uses only the *core* features (clearly known at creation). Set `INCLUDE_SNAPSHOT_FEATURES = True` to add the Tier 2 snapshot-risk features.

```python
INCLUDE_SNAPSHOT_FEATURES = False
MIN_FREQUENCY = 20

CORE_NUMERIC = [c for c in NUMERIC_FEATURES if c not in SNAPSHOT_RISK_FEATURES]
CORE_CATEGORICAL = [c for c in CATEGORICAL_FEATURES if c not in SNAPSHOT_RISK_FEATURES]

if INCLUDE_SNAPSHOT_FEATURES:
    USED_NUMERIC, USED_CATEGORICAL = NUMERIC_FEATURES, CATEGORICAL_FEATURES
else:
    USED_NUMERIC, USED_CATEGORICAL = CORE_NUMERIC, CORE_CATEGORICAL

print('Numeric features    :', USED_NUMERIC)
print('Categorical features:', USED_CATEGORICAL)
```

```python
def make_preprocessor(numeric_cols, categorical_cols):
    # Builds the preprocessing step for a given list of numeric and categorical columns
    numeric_pipe = Pipeline([
        ('impute', SimpleImputer(strategy='median', add_indicator=True)),
        ('scale', StandardScaler()),
    ])
    categorical_pipe = Pipeline([
        ('impute', SimpleImputer(strategy='constant', fill_value='Missing')),
        ('onehot', OneHotEncoder(handle_unknown='infrequent_if_exist', min_frequency=MIN_FREQUENCY,
                                 sparse_output=False)),
    ])
    return ColumnTransformer([
        ('num', numeric_pipe, numeric_cols),
        ('cat', categorical_pipe, categorical_cols),
    ])

preprocessor = make_preprocessor(USED_NUMERIC, USED_CATEGORICAL)
```

```python
# Look at what the preprocessor produces (fitted on TRAINING data only)
X_train_transformed = preprocessor.fit_transform(X_train)
feature_names_out = preprocessor.get_feature_names_out()
print('Shape after preprocessing:', X_train_transformed.shape)
print('Example output columns:', list(feature_names_out[:12]))
```

### Feature selection (training data only)

Some features may add nothing. Fewer inputs = a simpler API and less chance of fitting noise. To avoid peeking at the test set, selection uses **only the training set**:
1. Split the training set into 5 grouped folds (by student).
2. In each fold, fit a default gradient-boosting pipeline on 4 folds and compute **permutation importance** on the held-out fold; average over the 5 folds (one fold alone is too noisy).
3. Features whose average importance is ≤ `IMPORTANCE_CUTOFF` are candidates for removal.
4. Grouped 5-fold CV compares the full and reduced sets; the reduced set is kept if its PR-AUC is no more than `TOLERANCE` lower.

```python
# Grouped CV used for selection, model comparison and tuning
cv = StratifiedGroupKFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)
scoring = {'roc_auc': 'roc_auc', 'pr_auc': 'average_precision', 'brier': 'neg_brier_score'}

candidate_features = USED_NUMERIC + USED_CATEGORICAL
fold_importances = []

for fold, (fit_idx, val_idx) in enumerate(cv.split(X_train, y_train, groups_train), start=1):
    fs_model = Pipeline([('prep', make_preprocessor(USED_NUMERIC, USED_CATEGORICAL)),
                         ('model', HistGradientBoostingClassifier(learning_rate=0.05, max_iter=300, random_state=RANDOM_STATE))])
    fs_model.fit(X_train.iloc[fit_idx], y_train.iloc[fit_idx])
    perm_fold = permutation_importance(fs_model, X_train.iloc[val_idx][candidate_features], y_train.iloc[val_idx],
                                       scoring='average_precision', n_repeats=5, random_state=RANDOM_STATE, n_jobs=-1)
    fold_importances.append(pd.Series(perm_fold.importances_mean, index=candidate_features, name=f'fold {fold}'))
    print(f'fold {fold} done')

fs_importance = pd.concat(fold_importances, axis=1)
fs_importance['mean_drop_in_PR_AUC'] = fs_importance.mean(axis=1)
fs_importance = fs_importance.sort_values('mean_drop_in_PR_AUC', ascending=False)
fs_importance.round(4)
```

```python
IMPORTANCE_CUTOFF = 0.002
TOLERANCE = 0.005

LOW_VALUE_FEATURES = fs_importance.index[fs_importance['mean_drop_in_PR_AUC'] <= IMPORTANCE_CUTOFF].tolist()
REDUCED_NUMERIC = [c for c in USED_NUMERIC if c not in LOW_VALUE_FEATURES]
REDUCED_CATEGORICAL = [c for c in USED_CATEGORICAL if c not in LOW_VALUE_FEATURES]
print('Candidates for removal:', LOW_VALUE_FEATURES)

# Only test a reduced set if something would be removed AND something would remain
feature_sets = {'full feature set': (USED_NUMERIC, USED_CATEGORICAL)}
if LOW_VALUE_FEATURES and (REDUCED_NUMERIC or REDUCED_CATEGORICAL):
    feature_sets['reduced feature set'] = (REDUCED_NUMERIC, REDUCED_CATEGORICAL)
else:
    print('No usable reduced set (nothing to remove, or everything would be removed) -> full set kept.')

selection_rows = []
for name, (num_cols, cat_cols) in feature_sets.items():
    pipe = Pipeline([('prep', make_preprocessor(num_cols, cat_cols)),
                     ('model', HistGradientBoostingClassifier(learning_rate=0.05, max_iter=300, random_state=RANDOM_STATE))])
    scores = cross_validate(pipe, X_train, y_train, groups=groups_train, cv=cv, scoring=scoring, n_jobs=-1)
    selection_rows.append({'feature set': name, 'n_features': len(num_cols) + len(cat_cols),
                           'ROC-AUC': scores['test_roc_auc'].mean(), 'PR-AUC': scores['test_pr_auc'].mean(),
                           'PR-AUC std': scores['test_pr_auc'].std(), 'Brier': -scores['test_brier'].mean()})
selection_table = pd.DataFrame(selection_rows).set_index('feature set').round(4)
selection_table
```

```python
full_pr = selection_table.loc['full feature set', 'PR-AUC']
reduced_pr = selection_table['PR-AUC'].get('reduced feature set', np.nan)   # NaN if no reduced set was tested

if reduced_pr >= full_pr - TOLERANCE:
    USED_NUMERIC, USED_CATEGORICAL = REDUCED_NUMERIC, REDUCED_CATEGORICAL
    print(f'Reduced set kept (PR-AUC {reduced_pr:.4f} vs {full_pr:.4f}). Removed: {LOW_VALUE_FEATURES}')
else:
    print(f'Full set kept (PR-AUC {full_pr:.4f} vs reduced {reduced_pr:.4f}).')

print('Final numeric features    :', USED_NUMERIC)
print('Final categorical features:', USED_CATEGORICAL)
```

## Model comparison with grouped cross-validation

All models are compared with **5-fold `StratifiedGroupKFold` on the training set only** (the test set stays untouched).

| Model | Why it is included |
|---|---|
| Dummy (predicts the base rate) | The floor - any real model must beat it |
| Logistic regression (balanced) | Simple, interpretable linear baseline |
| Logistic regression (unweighted) | Same, with natural (uninflated) probabilities - the fairest "simple model" competitor |
| Random forest (balanced) | Non-linear, robust, handles interactions |
| Histogram gradient boosting | State-of-the-art for tabular data; native handling of many one-hot columns; fast |
| Histogram gradient boosting (balanced) | Same model with class weights; tests the **class-imbalance** strategy |

**Not used and why:** SMOTE/oversampling — creates synthetic applications that mix information across students (breaking the grouping), and distorts predicted probabilities which agents need to be meaningful; neural networks - small tabular dataset (~10k rows), where boosted trees are typically stronger and far easier to explain; SVM - no native probabilities and slow to tune at this size; XGBoost/LightGBM - `HistGradientBoostingClassifier` is the same family of algorithm without an extra dependency in the Docker image.

```python
# cv and scoring were defined in Section 9.1
candidate_models = {
    'Dummy (base rate)': DummyClassifier(strategy='prior'),
    'Logistic Regression (balanced)': LogisticRegression(max_iter=2000, class_weight='balanced'),
    'Logistic Regression (unweighted)': LogisticRegression(max_iter=2000),
    'Random Forest (balanced)': RandomForestClassifier(n_estimators=300, min_samples_leaf=5,
                                                       class_weight='balanced_subsample',
                                                       n_jobs=-1, random_state=RANDOM_STATE),
    'Hist Gradient Boosting': HistGradientBoostingClassifier(learning_rate=0.05, max_iter=300,
                                                             random_state=RANDOM_STATE),
    'Hist Gradient Boosting (balanced)': HistGradientBoostingClassifier(learning_rate=0.05, max_iter=300,
                                                                        class_weight='balanced',
                                                                        random_state=RANDOM_STATE),
}

cv_rows = []
for name, model in candidate_models.items():
    pipe = Pipeline([('prep', make_preprocessor(USED_NUMERIC, USED_CATEGORICAL)), ('model', model)])
    scores = cross_validate(pipe, X_train, y_train, groups=groups_train, cv=cv, scoring=scoring, n_jobs=-1)
    cv_rows.append({'model': name,
                    'ROC-AUC': scores['test_roc_auc'].mean(), 'ROC-AUC std': scores['test_roc_auc'].std(),
                    'PR-AUC': scores['test_pr_auc'].mean(), 'PR-AUC std': scores['test_pr_auc'].std(),
                    'Brier': -scores['test_brier'].mean()})
    print(f'done: {name}')

cv_results = pd.DataFrame(cv_rows).set_index('model').round(4)
cv_results.sort_values('PR-AUC', ascending=False)
```

```python
# Apply the choice rule: which models are tied with the best PR-AUC (within 1 std)?
best_name = cv_results['PR-AUC'].idxmax()
tie_limit = cv_results.loc[best_name, 'PR-AUC'] - cv_results.loc[best_name, 'PR-AUC std']
tied = cv_results[cv_results['PR-AUC'] >= tie_limit].sort_values('Brier')
print(f'Best PR-AUC: {best_name}. Models within 1 std of it, ordered by Brier (best first):')
tied[['PR-AUC', 'PR-AUC std', 'Brier']]
```

**How to read this:**
- A PR-AUC of the dummy model equals the base enrolment rate; the *lift* over it is what matters.
- Compare the balanced and unbalanced gradient boosting: class weights usually change **ROC/PR-AUC very little** (ranking is similar) but make the **Brier score much worse**, because the predicted probabilities are pushed upwards. Since agents need a *probability*, the unweighted model is preferred and imbalance is handled through the metric choice (PR-AUC) and the decision threshold instead.
- **Model choice rule:** models whose PR-AUC is within one standard deviation of the best are statistically tied on ranking. Among tied models, the **Brier score** (probability quality) decides, because agents read the output as a probability. If the unweighted logistic regression is tied on both, it is the better choice (simpler, explainable coefficients). Gradient boosting is tuned and the final choice between it and logistic regression is made after.

## Leakage & snapshot-feature experiments

Three versions of the same gradient-boosting model, evaluated with the same grouped CV:
1. **Core** features (default).
2. **Core + snapshot-risk** features (Tier 2).
3. **Core + leaky** milestone flags (Tier 1) - *deliberately wrong*, to show what leakage looks like.

```python
# Leaky flags built only for this demonstration (never used in the real model)
leaky_train = X_train.copy()
leaky_train['has_submission_date'] = model_df['Application Submission Date'].iloc[train_idx].notna().astype(int).values
leaky_train['has_offer_date'] = model_df['Offer Letter Received Date'].iloc[train_idx].notna().astype(int).values
leaky_train['has_deposit_date'] = model_df['Deposit Paid Date'].iloc[train_idx].notna().astype(int).values
LEAKY_FLAGS = ['has_submission_date', 'has_offer_date', 'has_deposit_date']

SNAP_NUMERIC = USED_NUMERIC + [c for c in SNAPSHOT_RISK_FEATURES if c in NUMERIC_FEATURES and c not in USED_NUMERIC]
SNAP_CATEGORICAL = USED_CATEGORICAL + [c for c in SNAPSHOT_RISK_FEATURES if c in CATEGORICAL_FEATURES and c not in USED_CATEGORICAL]

experiments = {
    '1. Selected features': (USED_NUMERIC, USED_CATEGORICAL),
    '2. Selected + snapshot-risk features': (SNAP_NUMERIC, SNAP_CATEGORICAL),
    '3. Selected + LEAKY milestone flags': (USED_NUMERIC + LEAKY_FLAGS, USED_CATEGORICAL),
}

exp_rows = []
for name, (num_cols, cat_cols) in experiments.items():
    pipe = Pipeline([('prep', make_preprocessor(num_cols, cat_cols)),
                     ('model', HistGradientBoostingClassifier(learning_rate=0.05, max_iter=300, random_state=RANDOM_STATE))])
    scores = cross_validate(pipe, leaky_train, y_train, groups=groups_train, cv=cv, scoring=scoring, n_jobs=-1)
    exp_rows.append({'experiment': name, 'ROC-AUC': scores['test_roc_auc'].mean(),
                     'PR-AUC': scores['test_pr_auc'].mean(), 'Brier': -scores['test_brier'].mean()})

pd.DataFrame(exp_rows).set_index('experiment').round(4)
```

**Interpretation guide:**
- **Experiment 3** should be close to perfect. That is *not* a good model - it simply reads the answer from the milestone columns. Reporting this kind of number would be the classic leakage mistake.
- **Experiment 2 vs 1:** the gain from snapshot-risk features is an **upper bound** on their true value, because some of it comes from fields being filled in *after* the application progressed. A small gain; keep the core model (safe). A large gain; the honest conclusion is that you need **historical snapshots of the CRM** (field values as of the scoring date) to use them safely.

## Hyperparameter tuning

Randomised search over the gradient-boosting hyperparameters, scored by **PR-AUC**, using the same grouped CV (so tuning cannot leak across students). Randomised search is used instead of a full grid because it covers a wide space with a fixed budget of fits.

**Grid design:** the grid deliberately extends to small leaves (`min_samples_leaf` 5), many iterations (up to 1000) and slow learning rates (0.01), so the best configuration is unlikely to sit at the edge of the search space. Built-in early stopping is **not** used: it holds out a *random* (not grouped-by-student) validation split internally, which would let the same student appear on both sides.

```python
tuning_pipeline = Pipeline([
    ('prep', make_preprocessor(USED_NUMERIC, USED_CATEGORICAL)),
    ('model', HistGradientBoostingClassifier(random_state=RANDOM_STATE)),
])

param_distributions = {
    'model__learning_rate': [0.01, 0.02, 0.05, 0.1],
    'model__max_iter': [150, 300, 500, 800, 1000],
    'model__max_leaf_nodes': [4, 8, 15, 31],
    'model__max_depth': [3, 5, None],
    'model__min_samples_leaf': [5, 10, 20, 50, 100],
    'model__l2_regularization': [0.0, 0.1, 1.0],
}

search = RandomizedSearchCV(tuning_pipeline, param_distributions, n_iter=40, scoring='average_precision',
                            cv=cv, n_jobs=-1, random_state=RANDOM_STATE, refit=True)
search.fit(X_train, y_train, groups=groups_train)

print('Best CV PR-AUC:', round(search.best_score_, 4))
print('Best parameters:')
for k, v in search.best_params_.items():
    print(f'   {k}: {v}')
```

**One-standard-error rule:** instead of the single best score, take every configuration whose CV PR-AUC is within one standard deviation of the best, and choose the **simplest** of them (fewest boosting iterations, then fewest leaves). Differences inside that band are noise, and simpler models generalise more reliably.

```python
results = pd.DataFrame(search.cv_results_)
best_row = results.loc[results['rank_test_score'] == 1].iloc[0]
within_one_std = results[results['mean_test_score'] >= best_row['mean_test_score'] - best_row['std_test_score']]

simplest = within_one_std.sort_values(['param_model__max_iter', 'param_model__max_leaf_nodes', 'mean_test_score'],
                                      ascending=[True, True, False]).iloc[0]
chosen_params = simplest['params']

print(f"Configurations within 1 std of the best: {len(within_one_std)} of {len(results)}")
print(f"Best CV PR-AUC     : {best_row['mean_test_score']:.4f}")
print(f"Chosen (simplest)  : {simplest['mean_test_score']:.4f} ± {simplest['std_test_score']:.4f}")
for k, v in chosen_params.items():
    print(f'   {k}: {v}')

# Refit the chosen configuration on the whole training set
tuned_model = clone(tuning_pipeline).set_params(**chosen_params).fit(X_train, y_train)
```

## Final model choice & calibration

```python
logistic_model = Pipeline([('prep', make_preprocessor(USED_NUMERIC, USED_CATEGORICAL)),
                           ('model', LogisticRegression(max_iter=2000))])

# Out-of-fold predicted probabilities for the training set (grouped CV)
oof_hgb = cross_val_predict(tuned_model, X_train, y_train, groups=groups_train, cv=cv,
                            method='predict_proba', n_jobs=-1)[:, 1]
oof_logistic = cross_val_predict(logistic_model, X_train, y_train, groups=groups_train, cv=cv,
                                 method='predict_proba', n_jobs=-1)[:, 1]

choice_table = pd.DataFrame({
    'ROC-AUC': [roc_auc_score(y_train, oof_hgb), roc_auc_score(y_train, oof_logistic)],
    'PR-AUC': [average_precision_score(y_train, oof_hgb), average_precision_score(y_train, oof_logistic)],
    'Brier': [brier_score_loss(y_train, oof_hgb), brier_score_loss(y_train, oof_logistic)],
}, index=['Tuned gradient boosting', 'Logistic regression']).round(4)
print('Base-rate Brier score:', round(brier_score_loss(y_train, np.full(len(y_train), y_train.mean())), 4))
choice_table
```

```python
PR_AUC_TOLERANCE = 0.02

logistic_tied = choice_table.loc['Logistic regression', 'PR-AUC'] >= choice_table.loc['Tuned gradient boosting', 'PR-AUC'] - PR_AUC_TOLERANCE
logistic_better_probs = choice_table.loc['Logistic regression', 'Brier'] <= choice_table.loc['Tuned gradient boosting', 'Brier']

if logistic_tied and logistic_better_probs:
    FINAL_ALGORITHM = 'Logistic regression'
    selected_model = logistic_model.fit(X_train, y_train)
    oof_proba = oof_logistic
else:
    FINAL_ALGORITHM = 'Tuned gradient boosting'
    selected_model = tuned_model
    oof_proba = oof_hgb
print('Selected model:', FINAL_ALGORITHM)

# Reliability diagram: points on the diagonal = well calibrated
fig, ax = plt.subplots(figsize=(7, 6))
for proba, label in [(oof_hgb, 'tuned gradient boosting'), (oof_logistic, 'logistic regression')]:
    frac_pos, mean_pred = calibration_curve(y_train, proba, n_bins=10, strategy='quantile')
    ax.plot(mean_pred, frac_pos, marker='o', label=label)
ax.plot([0, 1], [0, 1], 'k--', label='perfect calibration')
ax.set_xlabel('Mean predicted probability'); ax.set_ylabel('Observed enrolment rate')
ax.set_title('Calibration (out-of-fold, training set)'); ax.legend(); plt.show()
```

## Final evaluation on the held-out test set

Decision threshold and priority bands are chosen from the **training out-of-fold** predictions and never from the test set.

```python
# 1.Threshold that maximises F1 on the out-of-fold training predictions
precisions, recalls, thresholds = precision_recall_curve(y_train, oof_proba)
f1_scores = 2 * precisions[:-1] * recalls[:-1] / (precisions[:-1] + recalls[:-1] + 1e-12)
BEST_THRESHOLD = float(thresholds[np.argmax(f1_scores)])

# 2.Priority bands for agents: top 20% of scores = High, next 30% = Medium, rest = Low
HIGH_CUTOFF = float(np.quantile(oof_proba, 0.80))
MEDIUM_CUTOFF = float(np.quantile(oof_proba, 0.50))

print(f'F1-optimal threshold      : {BEST_THRESHOLD:.4f}')
print(f'High priority if score >= : {HIGH_CUTOFF:.4f}')
print(f'Medium priority if score >= {MEDIUM_CUTOFF:.4f}')
```

```python
# Predict on the untouched test set
test_proba = selected_model.predict_proba(X_test)[:, 1]
test_pred = (test_proba >= BEST_THRESHOLD).astype(int)

test_metrics = {
    'ROC-AUC': roc_auc_score(y_test, test_proba),
    'PR-AUC': average_precision_score(y_test, test_proba),
    'PR-AUC of random guessing (base rate)': y_test.mean(),
    'Brier score': brier_score_loss(y_test, test_proba),
    'Log loss': log_loss(y_test, test_proba),
    'F1 at chosen threshold': f1_score(y_test, test_pred),
}
pd.Series(test_metrics).round(4).to_frame('test set')
```

```python
# ROC and Precision-Recall curves
fpr, tpr, _ = roc_curve(y_test, test_proba)
prec, rec, _ = precision_recall_curve(y_test, test_proba)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
axes[0].plot(fpr, tpr, label=f"AUC = {test_metrics['ROC-AUC']:.3f}")
axes[0].plot([0, 1], [0, 1], 'k--'); axes[0].set_xlabel('False positive rate'); axes[0].set_ylabel('True positive rate')
axes[0].set_title('ROC curve (test)'); axes[0].legend()
axes[1].plot(rec, prec, label=f"PR-AUC = {test_metrics['PR-AUC']:.3f}")
axes[1].axhline(y_test.mean(), color='red', linestyle='--', label='base rate')
axes[1].set_xlabel('Recall'); axes[1].set_ylabel('Precision'); axes[1].set_title('Precision-Recall curve (test)')
axes[1].legend(); plt.tight_layout(); plt.show()
```

```python
# Confusion matrix and classification report at the chosen threshold
ConfusionMatrixDisplay(confusion_matrix(y_test, test_pred), display_labels=['not enrolled', 'enrolled']).plot(cmap='Blues')
plt.title(f'Confusion matrix (threshold = {BEST_THRESHOLD:.3f})'); plt.show()
print(classification_report(y_test, test_pred, target_names=['not enrolled', 'enrolled'], digits=3, zero_division=0))
```

**Lift / decile table** - the most business-friendly way to present the model: sort test applications by score, split into 10 equal groups and look at how many real enrolments each group contains.

```python
lift = pd.DataFrame({'score': test_proba, 'enrolled': y_test.values})
lift['decile'] = pd.qcut(lift['score'].rank(method='first', ascending=False), 10, labels=range(1, 11))

lift_table = lift.groupby('decile', observed=True).agg(applications=('enrolled', 'size'),
                                                       enrolments=('enrolled', 'sum'),
                                                       mean_score=('score', 'mean'))
lift_table['enrolment_rate_%'] = (lift_table['enrolments'] / lift_table['applications'] * 100).round(1)
lift_table['lift_vs_average'] = (lift_table['enrolment_rate_%'] / (y_test.mean() * 100)).round(2)
lift_table['cumulative_%_of_enrolments'] = (lift_table['enrolments'].cumsum() / lift_table['enrolments'].sum() * 100).round(1)
lift_table
```

```python
# Subgroup performance: does the model work equally well for different groups?
test_frame = X_test.copy()
test_frame['target'] = y_test.values
test_frame['score'] = test_proba

sub_rows = []
MIN_ENROLMENTS_FOR_RELIABLE_AUC = 20
for col in ['Lead Source Main Category', 'destination_type', 'level_of_study']:
    for value, g in test_frame.groupby(test_frame[col].fillna('Missing')):
        if g['target'].nunique() < 2 or len(g) < 50:
            continue   # AUC is undefined / unreliable for tiny or single-class groups
        sub_rows.append({'feature': col, 'group': value, 'n': len(g), 'enrolments': int(g['target'].sum()),
                         'ROC-AUC': roc_auc_score(g['target'], g['score']),
                         'mean_predicted_%': g['score'].mean() * 100, 'actual_rate_%': g['target'].mean() * 100,
                         'reliable': g['target'].sum() >= MIN_ENROLMENTS_FOR_RELIABLE_AUC})
pd.DataFrame(sub_rows).round(2)
```

## Temporal robustness check

The grouped random split answers "does the model generalise to **new students**?". In production it must also generalise to the **future**. Here we train on older applications and test on the most recent 20% (still with no student in both).

```python
time_cutoff = model_df['Created At'].quantile(0.80)
is_recent = model_df['Created At'] >= time_cutoff
recent_students = set(model_df.loc[is_recent, 'Student ID'])

temporal_train = model_df[~is_recent & ~model_df['Student ID'].isin(recent_students)]
temporal_test = model_df[is_recent]

temporal_model = clone(selected_model).fit(temporal_train[ALL_FEATURES], temporal_train['target'])
temporal_proba = temporal_model.predict_proba(temporal_test[ALL_FEATURES])[:, 1]

print(f'Time cutoff: {time_cutoff.date()} | train {len(temporal_train):,} rows, test {len(temporal_test):,} rows')
pd.DataFrame({
    'grouped random split': [test_metrics['ROC-AUC'], test_metrics['PR-AUC'], y_test.mean()],
    # temporal test set now only contains applications inside the training window (Section 4.2)
    'temporal split': [roc_auc_score(temporal_test['target'], temporal_proba),
                       average_precision_score(temporal_test['target'], temporal_proba),
                       temporal_test['target'].mean()],
}, index=['ROC-AUC', 'PR-AUC', 'base rate']).round(4)
```

A clear drop on the temporal split signals **drift** (the business changed over time); a strong argument for periodic retraining and drift monitoring (e.g. Evidently AI) in the deployed system. Because of the training window, the temporal test set contains only applications whose intake has passed, so recent censored applications do not distort this comparison.

### Limitations to discuss
- **Snapshot data**: fields hold their latest value, not the value at creation (residual leakage risk for profile fields). `Study Destination Type` and `Visa Is Required` were dropped for this reason; an undecided intake/destination may carry a similar, smaller effect.
- **Training window**: excluding applications whose intake has not passed removes right-censoring, but means the newest behaviour is learned last, therefore retrain regularly.
- **Multiple enrolments**: where one enrolled application has an `Enrolled Date` it is kept and the other enrolled copies removed; students with no or several dated enrolments are removed entirely. This loses some data and genuine transfers cannot be fully distinguished from CRM duplicates.
- **Raw-data quality**: junk dates, impossible dates of birth, contact details typed into the country field. All are handled and worth citing as evidence the data is raw.
- **Stale rule**: the 12-month rule for early-stage open applications is an assumption, show results with `STALE_MONTHS = None` as a sensitivity check.
- **Label definition**: "enrolled but dropped out / no commission" counted as success.
- **Deal Owner**: legitimate (no reassignment), but it may reflect lead allocation as well as skill; new agents fall into the *infrequent* bucket until their applications close.
- **Drift** from the temporal check
